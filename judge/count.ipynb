```python
import os
import pandas as pd

def count_a_b_in_file(file_path):
    with open(file_path, 'r') as file:
        content = file.read()
        count_a = content.count('A')
        count_b = content.count('B')
        return count_a, count_b

def count_a_b_in_directory(directory):
    total_a = 0
    total_b = 0
    for filename in os.listdir(directory):
        if filename.endswith('.txt'):
            file_path = os.path.join(directory, filename)
            count_a, count_b = count_a_b_in_file(file_path)
            total_a += count_a
            total_b += count_b
    return total_a, total_b
```

```python
# debate vs rubric
directory_path = '/Users/ylm/THU/code/exp/data/win/debate_vs_rubric'
a, b = count_a_b_in_directory(directory_path)
print('debate vs rubric: debate: {}, rubric: {}'.format(a, b-1))
print('win rate: {}'.format(a / (a + b-1)))
```

```python
# debate vs baseline
directory_path = '/Users/ylm/THU/code/exp/data/win/debate_vs_baseline'
a, b = count_a_b_in_directory(directory_path)
print('debate vs baseline: debate: {}, baseline: {}'.format(a, b-1))
print('win rate: {}'.format(a / (a + b-1)))
```
